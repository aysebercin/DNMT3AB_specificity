# ionic pairwise

```python
import seaborn as sns
from matplotlib import pyplot as plt
import pandas as pd
import csv

a=pd.read_csv(r"3A-CGC.int.ionics.inter", sep="\t", header=None )
a.columns = ["itype", "chain_a", "chain_b", "resname_a", "resname_b", "resid_a", "resid_b","atom_a", "atom_b" ]

a['resid_a'] = a['resid_a'] + 629

a["resname_a"] = a["resname_a"].astype(str) + a["resid_a"].astype(str)
a["resname_b"] = a["resname_b"].astype(str) + a["resid_b"].astype(str)

b=a.drop(columns=['resid_a','resid_b'])
b['Type'] = '3A:CGC'
```

```python
b['resname_b'] = b['resname_b'].str.replace('DA284',"A(+5')")
b['resname_b'] = b['resname_b'].str.replace('DG285',"G(+4')")
b['resname_b'] = b['resname_b'].str.replace('DA286',"A(+3')")
b['resname_b'] = b['resname_b'].str.replace('DG287',"G(+2')")
b['resname_b'] = b['resname_b'].str.replace('DC288',"C(+1')")
b['resname_b'] = b['resname_b'].str.replace('DG289',"G(0')")
b['resname_b'] = b['resname_b'].str.replace('DC290',"C(-1')")
b['resname_b'] = b['resname_b'].str.replace('DA291',"A(-2')")
b['resname_b'] = b['resname_b'].str.replace('DT292',"T(-3')")
b['resname_b'] = b['resname_b'].str.replace('DG293',"G(-4')")
b['resname_b'] = b['resname_b'].str.replace('DC294',"C(-4)")
b['resname_b'] = b['resname_b'].str.replace('DA295',"A(-3)")
b['resname_b'] = b['resname_b'].str.replace('DT296',"T(-2)")
b['resname_b'] = b['resname_b'].str.replace('DG297',"G(-1)")
b['resname_b'] = b['resname_b'].str.replace('DC298',"C(0)")
b['resname_b'] = b['resname_b'].str.replace('DG299',"G(+1)")
b['resname_b'] = b['resname_b'].str.replace('DC300',"C(+2)")
b['resname_b'] = b['resname_b'].str.replace('DT301',"T(+3)")
b['resname_b'] = b['resname_b'].str.replace('DC302',"C(+4)")
b['resname_b'] = b['resname_b'].str.replace('DT303',"T(+5)")


b['resname_a'] = b['resname_a'].str.title()
b
```

```python
b['pairwise'] = b['resname_a'].str.cat(b['resname_b'],sep="-")
b['pair_freq'] = b.groupby('pairwise')['pairwise'].transform('count')

#without flipped cytosine

b=b.drop(b.loc[b['resname_b']=='C(0)'].index)
b['aa_freq_wo_flip'] = b.groupby('resname_a')['resname_a'].transform('count')
b['pair_freq_wo_flip'] = b.groupby('pairwise')['pairwise'].transform('count')
b['nuc_freq_wo_flip'] = b.groupby('resname_b')['resname_b'].transform('count')
b['all_freq_wo_flip'] = b.groupby('itype')['itype'].transform('count')

b=b.sort_values(('pair_freq') , ascending=False)

b
```

```python
max_value = b['pair_freq_wo_flip'].max()

filter_b = b[b['pair_freq_wo_flip'] > (max_value*10)/100] 
```

```python
import seaborn as sns
from matplotlib import pyplot as plt
import pandas as pd
import csv


plt.figure(figsize=(10,10))
ax = sns.barplot(x="pairwise", y="pair_freq_wo_flip", palette="Blues_r", data=filter_b)
plt.xticks(rotation=30)


plt.xticks(fontsize=13, fontweight='bold')
plt.yticks(fontsize=13, fontweight='bold')
plt.ylim(ymax = 3000, ymin = 0)
plt.xlabel('Residues', fontsize=15, fontweight='bold')
plt.ylabel('Number of ionic bonds',fontsize=15, fontweight='bold')
plt.title("Interchain ionic bond profile - 3A:CGC",fontsize=20, fontweight='bold')

plt.savefig("3a-cgc-ionic-pairwise-filter.png", dpi=600, bbox_inches='tight', format="png")
```
