# Probability distributions of P-RMSD values during MD simulations

## DNMT3A

```python
import seaborn as sns
from matplotlib import pyplot as plt
import pandas as pd
import csv
import os
import warnings
warnings.filterwarnings('ignore')

input_files = os.path.join('DNMT3A_inputs')

a=pd.read_csv(input_files+"/3A-CGC-eq", sep="\t", header=None )
a.columns = ["frame", "rmsd"]
a['Type'] = 'CGC'

b=pd.read_csv(input_files+"/3A-CGG-eq", sep="\t", header=None )
b.columns = ["frame", "rmsd"]
b['Type'] = 'CGG'

c=pd.read_csv(input_files+"/3A-CGT-eq", sep="\t", header=None )
c.columns = ["frame", "rmsd"]
c['Type'] = 'CGT'

d=pd.read_csv(input_files+"/3A-CGA-eq", sep="\t", header=None )
d.columns = ["frame", "rmsd"]
d['Type'] = 'CGA'


rmsd_3a = pd.concat([a, b, c, d], ignore_index=True)

sns.set(style="ticks")

plt.figure(figsize=(6,6))

palet = {"CGC": "#C71585", "CGG": "#DCDCDC", "CGT": "orange", "CGA": "lightseagreen"}

sns.boxplot(y="rmsd", data=rmsd_3a, palette=palet,  x="Type", width=.5)

plt.xticks(fontsize=13)
plt.yticks(fontsize=13)
plt.ylim(ymax = 5, ymin = 1)

plt.ylabel('P-RMSD (Å)',fontsize=13)
plt.xlabel(' ')

plt.savefig("3a-prmsd.png", dpi=600, bbox_inches='tight', format="png")
```
