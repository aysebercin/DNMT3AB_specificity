# Pairwise upstream flanking nucleotide (-1 to 13) - residue mapping with h-bond and hydrophobic contact densities

```python
import os
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import csv
import warnings
warnings.filterwarnings('ignore')
```

## DNMT3A

```python
import seaborn as sns
from matplotlib import pyplot as plt
import pandas as pd
import numpy as np

sns.set(style="ticks")

input_files = os.path.join('DNMT3A_inputs')

#Please specify the DNA sequence motif here like cgc, cgg, cgt or cga
typ="CGC"

# File paths
hbond_target=pd.read_csv(input_files+"/heatmap_3a_"+typ+"_target_hbond", sep=" ")
hbond_compl=pd.read_csv(input_files+"/heatmap_3a_"+typ+"_compl_hbond", sep=" ")
hydrophobic_target=pd.read_csv(input_files+"/heatmap_3a_"+typ+"_target_hyd", sep=" ")
hydrophobic_compl=pd.read_csv(input_files+"/heatmap_3a_"+typ+"_compl_hyd", sep=" ")

# Normalize frequencies
hbond_target['frequency'] = hbond_target['pair_freq'] / 3204
hbond_compl['frequency'] = hbond_compl['pair_freq'] / 3204
hydrophobic_target['frequency'] = hydrophobic_target['pair_freq'] / 3204
hydrophobic_compl['frequency'] = hydrophobic_compl['pair_freq'] / 3204

# Pivot tables for heatmaps
hbond_target_pivot = hbond_target.pivot(index="resname_a", columns="resid_b", values="frequency")
hbond_compl_pivot = hbond_compl.pivot(index="resname_a", columns="resid_b", values="frequency")
hydrophobic_target_pivot = hydrophobic_target.pivot(index="resname_a", columns="resid_b", values="frequency")
hydrophobic_compl_pivot = hydrophobic_compl.pivot(index="resname_a", columns="resid_b", values="frequency")

# Reorder columns
column_order_t = ["(-3)", "(-2)", "(-1)"]
column_order_c = ["(-3')", "(-2')", "(-1')"]

hbond_target_pivot = hbond_target_pivot.reindex(column_order_t, axis=1)
hbond_compl_pivot = hbond_compl_pivot.reindex(column_order_c, axis=1)
hydrophobic_target_pivot = hydrophobic_target_pivot.reindex(column_order_t, axis=1)
hydrophobic_compl_pivot = hydrophobic_compl_pivot.reindex(column_order_c, axis=1)

# Sorting
hbond_compl_pivot.sort_values(by='resname_a', ascending=True, inplace=True)
hydrophobic_compl_pivot.sort_values(by='resname_a', ascending=True, inplace=True)

# Plot
fig, axes = plt.subplots(2, 2, figsize=(8, 8))

# Hydrogen bonds - Target
sns.heatmap(hbond_target_pivot, ax=axes[0, 0], cmap="RdPu", cbar=False,
            cbar_kws={'ticks': np.linspace(0, 1.8, 7)}, vmin=0, vmax=1.8,
            linewidths=0.1, linecolor='black', annot=True, annot_kws={'size': 10})
axes[0, 0].set_title("Hydrogen bonds")

# Hydrogen bonds - Complement
sns.heatmap(hbond_compl_pivot, ax=axes[1, 0], cmap="RdPu", cbar=False,
            cbar_kws={'ticks': np.linspace(0, 1.8, 7)}, vmin=0, vmax=1.8,
            linewidths=0.1, linecolor='black', annot=True, annot_kws={'size': 10})


# Hydrophobic - Target
sns.heatmap(hydrophobic_target_pivot, ax=axes[0, 1], cmap="bone_r", cbar=False,
            cbar_kws={'ticks': np.linspace(0, 1.5, 7)}, vmin=0, vmax=1.5,
            linewidths=0.1, linecolor='black', annot=True, annot_kws={'size': 10})
axes[0, 1].set_title("Hydrophobic interactions")

# Hydrophobic - Complement
sns.heatmap(hydrophobic_compl_pivot, ax=axes[1, 1], cmap="bone_r", cbar=False,
            cbar_kws={'ticks': np.linspace(0, 1.5, 7)}, vmin=0, vmax=1.5,
            linewidths=0.1, linecolor='black', annot=True, annot_kws={'size': 10})


# Formatting
for ax in axes.flat:
    tl = ax.get_xticklabels()
    ax.set_xticklabels(tl, fontsize=13, fontweight='bold')
    tly = ax.get_yticklabels()
    ax.set_yticklabels(tly, rotation=60, fontsize=14, fontweight='bold')
    ax.set_xlabel("")
    ax.set_ylabel("")

    for _, spine in ax.spines.items():
        spine.set_visible(True)

fig.suptitle("DNMT3A - "+typ, fontsize=20, fontweight='bold')

plt.tight_layout()

plt.savefig("heatmap-3a-"+typ+"_cb.pdf", dpi=600, bbox_inches='tight', format="pdf")

plt.show()
```
